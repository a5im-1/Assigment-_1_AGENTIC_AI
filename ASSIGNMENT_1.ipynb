```python
# Data types, variables, and operators

name = "Asim Ahmed"
age = 20
marks = 85.5
is_student = True

print("Name:", name)
print("Age:", age)
print("Marks:", marks)
print("Student:", is_student)

a = 10
b = 5

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Comparison:", a > b)
```

```python
# Control flow: if/else, loops, comprehensions

number = 7

if number % 2 == 0:
    print(number, "is even")
else:
    print(number, "is odd")
```

```python
for i in range(1, 6):
    print(i)
```

```python
count = 1

while count <= 5:
    print(count)
    count += 1
```

```python
numbers = [1, 2, 3, 4, 5]

squares = [x ** 2 for x in numbers]

print("Squares:", squares)
```
